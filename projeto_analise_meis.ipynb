# Projeto de Extensão
## Análise da Digitalização de Microempreendedores no Brasil

---

## 1. Diagnóstico e Teorização

### 1.1 Partes interessadas

As partes interessadas deste projeto são microempreendedores individuais (MEIs), como donos de farmácias, restaurantes e pequenos comércios locais, localizados no estado do Rio de Janeiro.

Esses empreendedores geralmente utilizam mais métodos tradicionais de venda e possuem pouca presença digital.

Além disso, também fazem parte do contexto os clientes desses estabelecimentos, que cada vez mais utilizam a internet para buscar produtos e serviços.

---

### 1.2 Problemática

Atualmente, muitos microempreendedores individuais (MEIs), como donos de pequenos comércios, farmácias e restaurantes, ainda possuem baixa presença digital.

Mesmo com o crescimento do uso da internet no Brasil, uma parte desses empreendedores não utiliza ferramentas digitais de forma estratégica, limitando sua visibilidade e alcance de clientes.

Segundo dados de instituições como o IBGE e o Sebrae (2024), os MEIs ainda apresentam um nível menor de digitalização quando comparados a empresas maiores, o que impacta diretamente na competitividade desses negócios.

Além disso, muitos desses empreendedores não utilizam ferramentas simples, como redes sociais ou plataformas de localização online, dificultando ainda mais a captação de novos clientes.

Diante disso, surge a necessidade de soluções que auxiliem esses pequenos negócios a entender e melhorar sua presença digital.

Outro ponto é o pouco conhecimento sobre tecnologias mais avançadas, como automações e inteligência artificial, que poderiam facilitar o atendimento e a gestão desses negócios.

---

### 1.3 Justificativa

Este projeto é importante pois está diretamente relacionado ao curso de Análise e Desenvolvimento de Sistemas (ADS), principalmente na área de análise de dados e uso de tecnologia para resolver problemas reais.

A proposta de analisar dados sobre microempreendedores permite aplicar, na prática, conhecimentos adquiridos durante o curso, como programação em Python e interpretação de dados.

Além disso, o projeto também possui relevância social, já que busca compreender as dificuldades enfrentadas por pequenos negócios em relação à presença digital, contribuindo para a inclusão digital e melhoria desses estabelecimentos.

Dessa forma, o projeto une teoria e prática, permitindo o desenvolvimento de habilidades técnicas ao mesmo tempo em que aborda um problema presente na sociedade.

Além disso, o projeto também considera o uso de tecnologias mais recentes, como automações e inteligência artificial, como possíveis soluções para auxiliar esses empreendedores.

---

### 1.4 Objetivos

- Analisar dados sobre microempreendedores individuais (MEIs), com foco na presença digital de pequenos negócios.

- Demonstrar como o uso de dados e ferramentas digitais pode ajudar esses empreendedores a melhorar sua visibilidade e atrair mais clientes.


---

### 1.5 Referencial Teórico

A transformação digital tem se tornado cada vez mais importante para o funcionamento e crescimento dos negócios na sociedade atual.

Segundo Manuel Castells, a sociedade contemporânea é marcada pelo uso intensivo de tecnologias digitais, o que influencia diretamente a forma como empresas e consumidores se relacionam.

Além disso, Philip Kotler destaca a importância do marketing digital e do uso de dados para entender o comportamento dos clientes, permitindo que empresas desenvolvam estratégias mais eficientes.

No contexto dos microempreendedores, a inclusão digital é um fator essencial para aumentar a competitividade e melhorar a gestão dos negócios.

Dessa forma, o uso de ferramentas tecnológicas e análise de dados se torna uma estratégia importante para auxiliar pequenos negócios a se adaptarem ao ambiente digital.
